Remove characters from the lyrics

```python
import re

newline = lambda x: re.sub('\n', ' ', x) # remove \n
alphanumeric = lambda x: re.sub('\w*\d\w*', ' ', x) # remove alphanumeric words

corpus = beatles.text.map(newline).map(alphanumeric)
corpus.head()
```

Create a document term matrix using Count Vectorizer with the stop words turned on to English

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(stop_words="english")
X = cv.fit_transform(corpus).toarray()

dt = pd.DataFrame(X, columns=cv.get_feature_names_out()).set_index(beatles.song)
dt.head()
```

Imagine lyrics in Count Vectorizer form


```python
imagine = list(dt.loc['Imagine'])
imagine[:20]
```

Define the cosine similarity calculation

```python
from numpy import dot
from numpy.linalg import norm

cosine = lambda v1, v2: dot(v1, v2) / (norm(v1) * norm(v2))
```

Calculate all similarities and sort by the most similar

```python
similarity = [cosine(imagine, song) for song in X]
sorted(list(zip(similarity, beatles.song)), reverse=True)[1:6]
```

Create the TF-IDF Vectorizer

```python
from sklearn.feature_extraction.text import TfidfVectorizer

cv_tfidf = TfidfVectorizer(stop_words="english")
X_tfidf = cv_tfidf.fit_transform(corpus).toarray()

dt_tfidf = pd.DataFrame(X_tfidf, columns=cv_tfidf.get_feature_names_out()).set_index(beatles.song)
dt_tfidf.head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

cv_tfidf = TfidfVectorizer(stop_words="english")
X_tfidf = cv_tfidf.fit_transform(corpus).toarray()

dt_tfidf = pd.DataFrame(X_tfidf, columns=cv_tfidf.get_feature_names_out()).set_index(beatles.song)
dt_tfidf.head()
```

Calculate all similarities and sort by the most similar using the TF-IDF data

```python
similarity_tfidf = [cosine(imagine, song) for song in X_tfidf]
sorted(list(zip(similarity_tfidf, beatles.song)), reverse=True)[1:6]
```

Use a dataset that consists of text document

```python
import pandas as pd

data = pd.read_csv('/content/drive/MyDrive/CPE 313/H.O.A. 8.1/Corona.csv', encoding='latin1')
data = data.drop(columns = ['ScreenName', 'Location', 'TweetAt'])
data.head()
```

Remove special characters

```python
import re
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

def filter_text(text):
    # remove URLs
    text = re.sub(r'http\S+|www\S+', '', text)
    # remove mentions and hashtags
    text = re.sub(r'@\w+|#\w+', '', text)
    # remove special characters and numbers
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    # lowercase
    text = text.lower()
    # strip extra spaces
    text = text.strip()
    # remove stopwords (optional)
    text = ' '.join([word for word in text.split() if word not in stop_words])
    return text

data['clean_tweet'] = data['OriginalTweet'].apply(filter_text)
data = data[data['clean_tweet'].str.strip() != '']
data.head()
```

```python
data_clean = data[['UserName', 'clean_tweet']].head(100)
data_text = data_clean['clean_tweet']
```

Compare the similar document using Count Vector and TF-IDF Vectorizer

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(stop_words='english')
x_cv = cv.fit_transform(data_text).toarray()

x_cv_df = pd.DataFrame(x_cv, columns=cv.get_feature_names_out(), index=data_clean['UserName'])
x_cv_df.head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tv = TfidfVectorizer(stop_words='english')
x_tv = tv.fit_transform(data_text).toarray()

x_tv_df = pd.DataFrame(x_tv, columns=tv.get_feature_names_out(), index=data_clean['UserName'])
x_tv_df.head()
```

```python
from numpy import dot
from numpy.linalg import norm

cosine = lambda v1, v2: dot(v1, v2) / (norm(v1) * norm(v2))
```

```python
user3800 = list(x_tv_df.iloc[0])
user3800[:20]
```

```python
similarity_count = [cosine (user3800, UserName) for UserName in x_cv]
sorted(list(zip(similarity_count, data_clean.UserName)), reverse=True)[1:6]
```

```python
similarity_count = [cosine (user3800, UserName) for UserName in x_tv]
sorted(list(zip(similarity_count, data_clean.UserName)), reverse=True)[1:6]
```

Use the cosine similarity to measure the similarity of text

```python
from itertools import combinations

pairs = list(combinations(data_clean.UserName.index, 2))
pairs_0 = list(combinations(range(len(data_clean)), 2))
covid_pairs = [(data_clean.UserName[a_index],data_clean.UserName[b_index]) for (a_index, b_index) in pairs]
covid_pairs[:5]
```

```python
results = [cosine(x_cv[a_index], x_cv[b_index]) for (a_index, b_index) in pairs_0]
sorted(zip(results, covid_pairs), reverse=True)[:5]
```

```python
results = [cosine(x_tv[a_index], x_tv[b_index]) for (a_index, b_index) in pairs_0]
sorted(zip(results, covid_pairs), reverse=True)[:5]
```

**Analysis:** The results show that Count Vectorizer gives higher similarity scores, especially when tweets share common words. For example, tweets 3821 and 3822 scored 0.935, which is very high. However, this method can be misleading because it treats all repeated words as equally important, even if the tweets have different meanings.
On the other hand, TF-IDF Vectorizer adjusts for word frequency, so common words have less influence. It still found tweets 3821 and 3822 to be very similar (score: 0.921), but other scores were lower and more balanced.
