```python
import pandas as pd

data = pd.read_csv('/content/drive/MyDrive/CPE 313/H.O.A. 8.1/Corona.csv', encoding='latin1')
data = data.drop(columns = ['ScreenName', 'Location', 'TweetAt'])
data.head()
```

```python
data = data[['UserName', 'OriginalTweet']].head(100)
data_text = data['OriginalTweet']
```

Tokenize (words)

```python
from nltk.tokenize import word_tokenize
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')

data['tokens'] = data_text.apply(lambda x: word_tokenize(x))
data.head()
```

Tokenize (words with remove punctuation) / Tokenize (words with remove special characters and numbers)

```python
import re
from nltk.tokenize import word_tokenize

# Remove punctuation
data['tokens_no_punct'] = data_text.apply(lambda x: word_tokenize(x.translate(str.maketrans('', '', string.punctuation))))

# Remove special characters & numbers
data['tokens_no_special'] = data_text.apply(lambda x: word_tokenize(re.sub(r'[^A-Za-z\s]', '', x)))

data.head()
```

Tokenize (words with remove punctuation and all lowercase)

```python
data['tokens_lower'] = data['tokens_no_punct'].apply(lambda x: [w.lower() for w in x])
data['tokens_lower'].head()
```

Tokenize (words with remove punctuation, all lowercase and remove stop words)

```python
from nltk.corpus import stopwords
nltk.download('stopwords')

stop_words = set(stopwords.words('english'))
data['tokens_no_stop'] = data['tokens_lower'].apply(lambda x: [w for w in x if w not in stop_words])
data['tokens_no_stop'].head()
```

Tokenize (words with remove punctuation, all lowercase and remove stop words, with speech_tagging)

```python
from nltk import pos_tag
nltk.download('averaged_perceptron_tagger_eng')

data['tokens_pos'] = data['tokens_no_stop'].apply(lambda x: nltk.pos_tag(x))
data['tokens_pos'].head()
```
